```python
import numpy as np
import pandas as pd
import sklearn
from sklearn.feature_extraction.text import CountVectorizer
```

```python
X_text=['This is my new bookshelf',
        'I really like it',
        'The old one got damaged',
        'This is very pretty',
        'I threw away the old one']
Y_text=['positive', 'positive', 'negative', 'positive', 'negative']
```

```python
vectorizer=CountVectorizer()
```

```python
vectorizer.fit(X_text)
```

```python
vocab=vectorizer.get_feature_names_out()
vocab
```

```python
X_term_doc_matrix=vectorizer.transform(X_text)
X_term_doc_matrix.toarray()
```

```python
df=pd.DataFrame(data=X_term_doc_matrix.toarray(),columns=vocab)
df
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
le.fit(Y_text)
y=le.transform(Y_text)
```

```python
import sklearn
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
X=X_term_doc_matrix
clf=model.fit(X,y)
```

```python
test=['like and pretty']
test_vector=vectorizer.transform(test)
print(test_vector.toarray())
df=pd.DataFrame(data=test_vector.toarray(),columns=vocab)
df
```

```python
pred=clf.predict(test_vector)
print(pred)
print(le.inverse_transform(pred))
```
